```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("diabetes_data.csv")

# Display first five records
print(df.head())

# Dataset size
print("Rows and columns:", df.shape)

# Column names and data types
print(df.dtypes)

# Statistical summary
print(df.describe())

# Check missing values
print(df.isnull().sum())

# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Correlation heatmap for numerical variables
plt.figure(figsize=(10, 7))
sns.heatmap(df.select_dtypes(include="number").corr(),
            annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap - Diabetes Lifestyle Dataset")
plt.show()

# Distribution of risk score
plt.figure(figsize=(8, 5))
plt.hist(df["risk_score"], bins=20, edgecolor="black")
plt.xlabel("Risk Score")
plt.ylabel("Frequency")
plt.title("Distribution of Diabetes Risk Score")
plt.show()

# Blood glucose vs risk score
plt.figure(figsize=(8, 5))
plt.scatter(df["blood_glucose"], df["risk_score"], alpha=0.6)
plt.xlabel("Blood Glucose")
plt.ylabel("Risk Score")
plt.title("Blood Glucose vs Risk Score")
plt.show()
```
