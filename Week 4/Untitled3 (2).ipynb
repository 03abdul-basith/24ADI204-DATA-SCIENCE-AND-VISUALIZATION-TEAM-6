```python
# 9. Stacked Bar Plot: 'diet' vs 'stress_level'
cross_tab_diet_stress = pd.crosstab(df_clean['diet'], df_clean['stress_level'], normalize='index')
cross_tab_diet_stress.plot(kind='bar', stacked=True, figsize=(8, 6))
plt.title('Proportion of Stress Levels by Diet')
plt.xlabel('Diet (0=No, 1=Yes)')
plt.ylabel('Proportion')
plt.xticks(rotation=0)
plt.legend(title='Stress Level', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# 10. Stacked Bar Plot: 'medication_adherence' vs 'hydration_level'
cross_tab_med_hydration = pd.crosstab(df_clean['medication_adherence'], df_clean['hydration_level'], normalize='index')
cross_tab_med_hydration.plot(kind='bar', stacked=True, figsize=(8, 6))
plt.title('Proportion of Hydration Levels by Medication Adherence')
plt.xlabel('Medication Adherence (0=No, 1=Yes)')
plt.ylabel('Proportion')
plt.xticks(rotation=0)
plt.legend(title='Hydration Level', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
